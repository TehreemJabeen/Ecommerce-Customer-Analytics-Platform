I identified negative quantities which represented returns and cancellations. Since my analysis focuses on purchase behavior for customer segmentation, I removed these transactions to avoid double-counting. The original purchase transactions were already captured with positive quantities. This is standard practice in retail analytics as returns represent a different business process.

```python
def identify_non_product_items(df):
    """
    Identify non-product items using exact word matching.
    Returns a boolean mask (True = non-product item).
    """
    
    # Known product descriptions that should NOT be flagged
    # (these contain keywords but are real products)
    product_exclusions = [
        'BROWN CHECK CAT DOORSTOP',
        'PAIR PADDED HANGERS PINK CHECK',
        'SUNSET CHECK HAMMOCK',
        'BLACK BAROQUE CARRIAGE CLOCK',
        'WHITE BAROQUE CARRIAGE CLOCK',
        'MULTICOLOUR CRUSOE CHECK LAMPSHADE',
        'BLUE CHECK BAG W HANDLE 34X20CM',
        'BLUE CRUSOE CHECK LAMPSHADE',
        'GOTHIC CARRIAGE LANTERN',
    ]
    
    # Normalize exclusions once for fast lookup
    product_exclusions_upper = {p.upper().strip() for p in product_exclusions}

    exact_word_keywords = [
        'POSTAGE',
        'CARRIAGE',
        'DISCOUNT',
        'MANUAL',
        'ADJUST',
        'ADJUSTMENT',
        'SAMPLES',
        'SAMPLE',
        'FEE',
        'FEES',
        'TEST',
        'CHECK',
        'DOTCOM',
    ]
    
    phrase_keywords = [
        'BANK CHARGES',
        'BANK CHARGE',
        'AMAZON FEE',
        'EBAY FEE',
        'DOTCOM POSTAGE',
        'CRUSHED',
        'SMASHED',
        'DAMAGED',
        'WRONG',
    ]
    
    def is_non_product(description):
        if pd.isna(description):
            return False
        
        desc_upper = str(description).upper().strip()
        
        # ✅ Exclusion check first — known real products are never flagged
        if desc_upper in product_exclusions_upper:
            return False
        
        # Check exact word matches
        for keyword in exact_word_keywords:
            pattern = r'\b' + re.escape(keyword) + r'\b'
            if re.search(pattern, desc_upper):
                return True
        
        # Check phrase matches
        for phrase in phrase_keywords:
            if phrase in desc_upper:
                return True
        
        return False
    
    mask = df['Description'].apply(is_non_product)
    return mask
```

```python
def clean_retail_data(df):
    """
    Clean the retail dataset.
    
    Returns: cleaned dataframe
    """
    
    print("Starting data cleaning...")
    print(f"Original shape: {df.shape}")
    print("=" * 60)
    
    df_clean = df.copy()
    
    # 1. Remove rows without Customer ID
    df_clean = df_clean[df_clean['Customer ID'].notna()]
    print(f"After removing missing Customer ID: {df_clean.shape} "
          f"({len(df) - len(df_clean):,} rows removed)")
    
    # 2. Remove cancelled orders (Invoice starting with 'C')
    df_clean = df_clean[~df_clean['Invoice'].astype(str).str.startswith('C')]
    print(f"After removing cancelled orders: {df_clean.shape} "
          f"({len(df) - len(df_clean):,} rows removed total)")
    
    # 3. Remove negative quantities
    df_clean = df_clean[df_clean['Quantity'] > 0]
    print(f"After removing negative quantities: {df_clean.shape}")
    
    # 4. Remove negative and zero prices
    df_clean = df_clean[df_clean['Price'] > 0]
    print(f"After removing negative/zero prices: {df_clean.shape}")
    
    # 5. Remove non-product items using identify_non_product_items()
    non_product_mask = identify_non_product_items(df_clean)
    rows_before = len(df_clean)
    df_clean = df_clean[~non_product_mask]
    print(f"After removing non-product items: {df_clean.shape} "
          f"({rows_before - len(df_clean):,} rows removed)")
    
    # 6. Remove rows with missing Description
    df_clean = df_clean[df_clean['Description'].notna()]
    print(f"After removing missing descriptions: {df_clean.shape}")
    
    # 7. Create TotalPrice column
    df_clean['TotalPrice'] = df_clean['Quantity'] * df_clean['Price']
    
    # 8. Keep date features if they exist
    date_features = ['Year', 'Month', 'Day', 'DayOfWeek', 'Hour', 'DayName', 'MonthName']
    existing_features = [f for f in date_features if f in df_clean.columns]
    
    if not existing_features:
        df_clean['Year'] = df_clean['InvoiceDate'].dt.year
        df_clean['Month'] = df_clean['InvoiceDate'].dt.month
        df_clean['Day'] = df_clean['InvoiceDate'].dt.day
        df_clean['DayOfWeek'] = df_clean['InvoiceDate'].dt.dayofweek
        df_clean['Hour'] = df_clean['InvoiceDate'].dt.hour
    
    # 9. Reset index
    df_clean = df_clean.reset_index(drop=True)
    
    print("=" * 60)
    print(f"✅ Cleaning complete!")
    print(f"Final shape: {df_clean.shape}")
    print(f"Total rows removed: {len(df) - len(df_clean):,} ({(len(df) - len(df_clean))/len(df)*100:.1f}%)")
    print(f"Rows retained: {len(df_clean):,} ({len(df_clean)/len(df)*100:.1f}%)")
    
    return df_clean
```

```python
df_clean=clean_retail_data(df)
```
